## Exercise 1: Tic-Tac-Toe AI

**AI = X (MAX)** | **Opponent = O (MIN)** | Board: 3×3 (flat list of 9) | X goes first

### Part A: Core Functions

```python
def initial_state():
    return [' '] * 9

def print_board(state):
    print()
    for row in range(3):
        cells = state[row*3 : row*3+3]
        print(f" {cells[0]} | {cells[1]} | {cells[2]} ")
        if row < 2:
            print("---+---+---")
    print()

def actions(state):
    return [i for i, c in enumerate(state) if c == ' ']

def result(state, action, mark):
    new = state[:]
    new[action] = mark
    return new

def player(state):
    return 'X' if state.count('X') == state.count('O') else 'O'

WIN_LINES = [
    [0,1,2],[3,4,5],[6,7,8],
    [0,3,6],[1,4,7],[2,5,8],
    [0,4,8],[2,4,6]
]

def utility(state):
    for line in WIN_LINES:
        vals = [state[i] for i in line]
        if vals == ['X','X','X']: return +1
        if vals == ['O','O','O']: return -1
    if ' ' not in state: return 0
    return None

def terminal(state):
    return utility(state) is not None or len(actions(state)) == 0

def game_status(state):
    u = utility(state)
    return {1: 'X wins!', -1: 'O wins!', 0: 'Draw!'}.get(u, 'Continue')

board = initial_state()
print("Initial Board:")
print_board(board)
print("Turn:", player(board))
print("Available actions:", actions(board))
```

### Part B: Minimax

```python
nodes_minimax = 0

def max_value(state):
    global nodes_minimax
    nodes_minimax += 1
    if terminal(state):
        return utility(state)
    v = float('-inf')
    for a in actions(state):
        v = max(v, min_value(result(state, a, 'X')))
    return v

def min_value(state):
    global nodes_minimax
    nodes_minimax += 1
    if terminal(state):
        return utility(state)
    v = float('inf')
    for a in actions(state):
        v = min(v, max_value(result(state, a, 'O')))
    return v

def minimax_decision(state):
    best_move, best_val = None, float('-inf')
    for a in actions(state):
        val = min_value(result(state, a, 'X'))
        if val > best_val:
            best_val, best_move = val, a
    return best_move
```

### Part C: Alpha–Beta Pruning

```python
nodes_ab = 0

def max_value_ab(state, alpha, beta):
    global nodes_ab
    nodes_ab += 1
    if terminal(state):
        return utility(state)
    v = float('-inf')
    for a in actions(state):
        v = max(v, min_value_ab(result(state, a, 'X'), alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def min_value_ab(state, alpha, beta):
    global nodes_ab
    nodes_ab += 1
    if terminal(state):
        return utility(state)
    v = float('inf')
    for a in actions(state):
        v = min(v, max_value_ab(result(state, a, 'O'), alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v

def alphabeta_decision(state):
    best_move, best_val = None, float('-inf')
    alpha, beta = float('-inf'), float('inf')
    for a in actions(state):
        val = min_value_ab(result(state, a, 'X'), alpha, beta)
        if val > best_val:
            best_val, best_move = val, a
        alpha = max(alpha, best_val)
    return best_move
```

### Game Loop: Human (O) vs AI (X)

```python
def play_tictactoe(use_alphabeta=True):
    global nodes_minimax, nodes_ab
    nodes_minimax = 0
    nodes_ab = 0

    state = initial_state()
    print("=== Tic-Tac-Toe: Human (O) vs AI (X) ===")
    print_board(state)

    while not terminal(state):
        current = player(state)

        if current == 'X':
            move = alphabeta_decision(state) if use_alphabeta else minimax_decision(state)
            row, col = divmod(move, 3)
            print(f"AI (X) plays at index {move} → (row {row}, col {col})")
            state = result(state, move, 'X')
        else:
            print_board(state)
            while True:
                try:
                    move = int(input("Your move (0-8): "))
                    if move in actions(state):
                        break
                    print("Cell taken or invalid. Try again.")
                except ValueError:
                    print("Enter a number 0-8.")
            state = result(state, move, 'O')

        print_board(state)
        print("Status:", game_status(state))

    print("\n=== Game Over ===")
    print(f"Nodes evaluated — Minimax: {nodes_minimax} | Alpha-Beta: {nodes_ab}")

play_tictactoe(use_alphabeta=True)
```

### Node Count Comparison (from empty board)

```python
nodes_minimax = 0
nodes_ab = 0

minimax_decision(initial_state())
alphabeta_decision(initial_state())

print(f"Nodes evaluated using Minimax    : {nodes_minimax}")
print(f"Nodes evaluated using Alpha-Beta : {nodes_ab}")
print(f"Reduction                        : {round((1 - nodes_ab/nodes_minimax)*100, 1)}%")
```

**Why Alpha–Beta evaluates fewer nodes:**  
Alpha–Beta pruning skips branches that cannot influence the final decision. When a MIN node finds a value ≤ α or a MAX node finds a value ≥ β, the remaining sibling branches are pruned entirely. This eliminates redundant subtrees while guaranteeing the exact same optimal move as plain Minimax.

## Exercise 2: Car Overtaking Game

**Road:** 7 cells (0–6) | **MAX (X)** starts at cell 0, **MIN (O)** starts at cell 1  
**Moves:** forward 1 or 2 cells | **Win:** first to reach cell 6

### Part A: Problem Formulation

```python
ROAD_SIZE = 7
GOAL = ROAD_SIZE - 1

def car_initial_state():
    return (0, 1)

def car_actions(state, is_max):
    pos = state[0] if is_max else state[1]
    opp = state[1] if is_max else state[0]
    return [s for s in [1, 2] if pos + s <= GOAL and pos + s != opp]

def car_result(state, move, is_max):
    x, o = state
    return (x + move, o) if is_max else (x, o + move)

def car_terminal(state):
    return state[0] == GOAL or state[1] == GOAL

def car_utility(state):
    if state[0] == GOAL: return +1
    if state[1] == GOAL: return -1
    return None

def display_road(state):
    road = ['_'] * ROAD_SIZE
    road[state[0]] = 'X'
    road[state[1]] = 'O'
    print(' '.join(road))
    print(f"  X at cell {state[0]}, O at cell {state[1]}")

display_road(car_initial_state())
```

### Part B: Minimax

```python
car_nodes_minimax = 0

def car_max_value(state):
    global car_nodes_minimax
    car_nodes_minimax += 1
    if car_terminal(state):
        return car_utility(state)
    v = float('-inf')
    for move in car_actions(state, True):
        v = max(v, car_min_value(car_result(state, move, True)))
    return v

def car_min_value(state):
    global car_nodes_minimax
    car_nodes_minimax += 1
    if car_terminal(state):
        return car_utility(state)
    v = float('inf')
    for move in car_actions(state, False):
        v = min(v, car_max_value(car_result(state, move, False)))
    return v

def car_minimax_decision(state):
    best_move, best_val = None, float('-inf')
    for move in car_actions(state, True):
        val = car_min_value(car_result(state, move, True))
        if val > best_val:
            best_val, best_move = val, move
    return best_move
```

### Part C: Alpha–Beta Pruning

```python
car_nodes_ab = 0

def car_max_value_ab(state, alpha, beta):
    global car_nodes_ab
    car_nodes_ab += 1
    if car_terminal(state):
        return car_utility(state)
    v = float('-inf')
    for move in car_actions(state, True):
        v = max(v, car_min_value_ab(car_result(state, move, True), alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def car_min_value_ab(state, alpha, beta):
    global car_nodes_ab
    car_nodes_ab += 1
    if car_terminal(state):
        return car_utility(state)
    v = float('inf')
    for move in car_actions(state, False):
        v = min(v, car_max_value_ab(car_result(state, move, False), alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v

def car_alphabeta_decision(state):
    best_move, best_val = None, float('-inf')
    alpha, beta = float('-inf'), float('inf')
    for move in car_actions(state, True):
        val = car_min_value_ab(car_result(state, move, True), alpha, beta)
        if val > best_val:
            best_val, best_move = val, move
        alpha = max(alpha, best_val)
    return best_move
```

### Game Simulation: AI (X) vs AI (O)

```python
def play_car_game(use_alphabeta=True):
    global car_nodes_minimax, car_nodes_ab
    car_nodes_minimax = 0
    car_nodes_ab = 0

    state = car_initial_state()
    is_max_turn = True
    turn = 0

    print(f"=== Car Overtaking Game | Road: {ROAD_SIZE} cells (0 to {GOAL}) ===")
    print()

    while not car_terminal(state):
        display_road(state)
        label = 'X (MAX)' if is_max_turn else 'O (MIN)'

        if is_max_turn:
            move = car_alphabeta_decision(state) if use_alphabeta else car_minimax_decision(state)
            state = car_result(state, move, True)
        else:
            moves = car_actions(state, False)
            move = moves[0] if moves else None
            if move:
                state = car_result(state, move, False)

        print(f"Turn {turn+1}: {label} moves forward by {move}")
        print(f"Updated → X at {state[0]}, O at {state[1]}")
        print()
        is_max_turn = not is_max_turn
        turn += 1

    display_road(state)
    outcome = 'X wins!' if state[0] == GOAL else 'O wins!'
    print(f"\n=== {outcome} ===")
    print(f"Nodes evaluated — Minimax: {car_nodes_minimax} | Alpha-Beta: {car_nodes_ab}")

play_car_game(use_alphabeta=True)
```

### Node Count Comparison

```python
car_nodes_minimax = 0
car_nodes_ab = 0

car_minimax_decision(car_initial_state())
car_alphabeta_decision(car_initial_state())

print(f"Nodes evaluated using Minimax    : {car_nodes_minimax}")
print(f"Nodes evaluated using Alpha-Beta : {car_nodes_ab}")
if car_nodes_minimax > 0:
    print(f"Reduction                        : {round((1 - car_nodes_ab/car_nodes_minimax)*100, 1)}%")
```

### Extension: Longer Road + Obstacles + Depth-Limited Search

```python
def play_car_extended(road_size=12, obstacles=None, max_depth=10):
    if obstacles is None:
        obstacles = set()
    goal = road_size - 1

    def ext_actions(state, is_max):
        pos = state[0] if is_max else state[1]
        opp = state[1] if is_max else state[0]
        return [s for s in [1, 2] if pos+s <= goal and pos+s != opp and pos+s not in obstacles]

    def ext_result(state, move, is_max):
        x, o = state
        return (x+move, o) if is_max else (x, o+move)

    def ext_terminal(state):
        return state[0] == goal or state[1] == goal

    def ext_utility(state):
        if state[0] == goal: return +1
        if state[1] == goal: return -1
        return None

    def heuristic(state):
        return (state[0] - state[1]) / goal

    def ext_max_ab(state, alpha, beta, depth):
        if ext_terminal(state):
            return ext_utility(state)
        if depth == max_depth:
            return heuristic(state)
        v = float('-inf')
        for move in ext_actions(state, True):
            v = max(v, ext_min_ab(ext_result(state, move, True), alpha, beta, depth+1))
            if v >= beta: return v
            alpha = max(alpha, v)
        return v

    def ext_min_ab(state, alpha, beta, depth):
        if ext_terminal(state):
            return ext_utility(state)
        if depth == max_depth:
            return heuristic(state)
        v = float('inf')
        for move in ext_actions(state, False):
            v = min(v, ext_max_ab(ext_result(state, move, False), alpha, beta, depth+1))
            if v <= alpha: return v
            beta = min(beta, v)
        return v

    def ext_decision(state):
        moves = ext_actions(state, True)
        if not moves: return None
        best_move, best_val = moves[0], float('-inf')
        alpha, beta = float('-inf'), float('inf')
        for move in moves:
            val = ext_min_ab(ext_result(state, move, True), alpha, beta, 0)
            if val >= best_val:
                best_val, best_move = val, move
            alpha = max(alpha, best_val)
        return best_move

    state = (0, 1)
    is_max_turn = True
    print(f"=== Extended | Road: {road_size} cells | Obstacles: {obstacles} | Max depth: {max_depth} ===")

    while not ext_terminal(state):
        road = ['_'] * road_size
        for ob in obstacles: road[ob] = '#'
        road[state[0]] = 'X'
        road[state[1]] = 'O'
        print(' '.join(road))

        if is_max_turn:
            move = ext_decision(state)
            if move is None:
                print('X is blocked!')
                break
            state = ext_result(state, move, True)
            print(f"X moves +{move} → X:{state[0]}, O:{state[1]}")
        else:
            moves = ext_actions(state, False)
            if not moves:
                print('O is blocked!')
                break
            move = moves[0]
            state = ext_result(state, move, False)
            print(f"O moves +{move} → X:{state[0]}, O:{state[1]}")
        is_max_turn = not is_max_turn
        print()

    outcome = 'X wins!' if state[0] == goal else 'O wins!'
    print(f"=== {outcome} ===")

play_car_extended(road_size=12, obstacles={5, 9}, max_depth=8)
```
