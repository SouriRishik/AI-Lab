## 1) Smart Lighting Logical Agent

```python
class SmartLightingAgent:
    def __init__(self):
        self.kb = {}

    def update_kb(self, motion, dark, override):
        self.kb["MotionDetected"] = motion
        self.kb["IsDark"] = dark
        self.kb["ManualOverride"] = override

    def infer(self):
        if self.kb["ManualOverride"]:
            return "No Action"
        if self.kb["IsDark"] and self.kb["MotionDetected"]:
            return "Light ON"
        if not self.kb["MotionDetected"]:
            return "Light OFF"
        return "No Action"

agent = SmartLightingAgent()
agent.update_kb(True, True, False)
print(agent.infer())
```

## 2) Medical Diagnosis Logical Agent

```python
class MedicalAgent:
    def __init__(self):
        self.facts = set()

    def add_symptoms(self, fever, cough, chestpain):
        if fever:
            self.facts.add("Fever")
        if cough:
            self.facts.add("Cough")
        if chestpain:
            self.facts.add("ChestPain")

    def forward_chain(self):
        if "Fever" in self.facts and "Cough" in self.facts:
            self.facts.add("Flu")
        if "Fever" in self.facts and "Cough" in self.facts and "ChestPain" in self.facts:
            self.facts.add("Pneumonia")
        if "Pneumonia" in self.facts:
            self.facts.add("HospitalizationRequired")
        return self.facts

agent = MedicalAgent()
agent.add_symptoms(True, True, True)
print(agent.forward_chain())
```

## 3) Traffic Violation Logical Agent

```python
class TrafficAgent:
    def __init__(self):
        self.kb = {}

    def perceive(self, speed, red, crossed, emergency):
        self.kb["SpeedAboveLimit"] = speed
        self.kb["SignalRed"] = red
        self.kb["VehicleCrossedLine"] = crossed
        self.kb["EmergencyVehicle"] = emergency

    def infer(self):
        violations = []
        if not self.kb["EmergencyVehicle"]:
            if self.kb["SpeedAboveLimit"]:
                violations.append("SpeedViolation")
            if self.kb["SignalRed"] and self.kb["VehicleCrossedLine"]:
                violations.append("SignalViolation")
        return violations

    def act(self):
        v = self.infer()
        if v:
            return "Fine Issued: " + ", ".join(v)
        return "No Action"

agent = TrafficAgent()
agent.perceive(True, True, True, False)
print(agent.act())
```

## 4) Cybersecurity Intrusion Detection Agent

```python
class SecurityAgent:
    def __init__(self):
        self.facts = set()

    def update(self, failed, unknown_ip, sensitive, admin):
        if failed:
            self.facts.add("MultipleFailedLogins")
        if unknown_ip:
            self.facts.add("LoginFromUnknownIP")
        if sensitive:
            self.facts.add("AccessToSensitiveFiles")
        if admin:
            self.facts.add("AdminPrivileges")

    def infer(self):
        if "MultipleFailedLogins" in self.facts and "LoginFromUnknownIP" in self.facts:
            self.facts.add("SuspiciousLogin")
        if "SuspiciousLogin" in self.facts and "AdminPrivileges" in self.facts:
            self.facts.add("HighRiskIntrusion")
        if "HighRiskIntrusion" in self.facts:
            self.facts.add("TriggerAlert")
        return self.facts

    def act(self):
        f = self.infer()
        if "TriggerAlert" in f:
            return "Alert Triggered"
        return "System Normal"

agent = SecurityAgent()
agent.update(True, True, True, True)
print(agent.act())
```

## 5) Autonomous Warehouse Robot Logical Agent

```python
class WarehouseAgent:
    def __init__(self):
        self.kb = {}

    def perceive(self, obstacle, clear, fragile, human, reached):
        self.kb["ObstacleAhead"] = obstacle
        self.kb["PathClear"] = clear
        self.kb["CarryingFragileItem"] = fragile
        self.kb["HumanNearby"] = human
        self.kb["DestinationReached"] = reached

    def infer(self):
        if self.kb["DestinationReached"]:
            return "Stop"
        if self.kb["CarryingFragileItem"] and self.kb["ObstacleAhead"]:
            return "SlowDown"
        if self.kb["ObstacleAhead"]:
            return "Stop"
        if self.kb["HumanNearby"]:
            return "ReduceSpeed"
        if self.kb["PathClear"] and not self.kb["HumanNearby"]:
            return "MoveForward"
        return "Idle"

agent = WarehouseAgent()
agent.perceive(True, False, True, False, False)
print(agent.infer())
```
