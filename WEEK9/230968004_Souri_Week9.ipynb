### 1. Logic Engine

```python
import itertools

class ResolutionEngine:
    def __init__(self):
        self.kb = []

    def to_cnf(self, rule):
        if "->" in rule:
            antecedent, consequent = rule.split("->")
            antecedent = antecedent.strip()
            consequent = consequent.strip()
            
            if "∧" in antecedent:
                parts = antecedent.split("∧")
                clause = set()
                for p in parts:
                    p = p.strip()
                    clause.add(f"!{p}" if not p.startswith("!") else p[1:])
                clause.add(consequent)
                return [clause]
            elif "∨" in antecedent:
                parts = antecedent.split("∨")
                return [{f"!{p.strip()}", consequent} for p in parts]
            else:
                return [{f"!{antecedent}", consequent}]
        return [{rule.strip()}]

    def resolve(self, ci, cj):
        resolvents = []
        for di in ci:
            for dj in cj:
                if di == f"!{dj}" or dj == f"!{di}":
                    new_clause = (ci - {di}) | (cj - {dj})
                    resolvents.append(new_clause)
        return resolvents

    def prove(self, facts, rules, query):
        self.kb = []
        for r in rules:
            self.kb.extend(self.to_cnf(r))
        for f in facts:
            self.kb.append({f})
        
        print(f"CNF Clauses: {self.kb}")
        
        neg_query = f"!{query}" if not query.startswith("!") else query[1:]
        clauses = list(self.kb)
        clauses.append({neg_query})
        
        while True:
            new_clauses_found = False
            pairs = list(itertools.combinations(range(len(clauses)), 2))
            for (i, j) in pairs:
                resolvents = self.resolve(clauses[i], clauses[j])
                for res in resolvents:
                    if not res:
                        return True
                    if res not in clauses:
                        clauses.append(res)
                        new_clauses_found = True
            if not new_clauses_found:
                return False

def run_test(name, facts, rules, query):
    print(f"\n--- {name} ---")
    engine = ResolutionEngine()
    result = engine.prove(facts, rules, query)
    print(f"Result: {query} = {result}")
```

### 2. Cybersecurity Threat Detection

```python
cyber_rules = [
    "MultipleFailedLogins ∧ LoginFromUnknownIP -> SuspiciousLogin",
    "SuspiciousLogin ∧ AdminPrivileges -> HighRiskIntrusion",
    "HighRiskIntrusion -> IntrusionDetected",
    "AccessToSensitiveFiles ∧ !AdminPrivileges -> IntrusionDetected"
]

run_test("Cyber Scenario 1", ["MultipleFailedLogins"], cyber_rules, "IntrusionDetected")
run_test("Cyber Scenario 2", ["MultipleFailedLogins", "LoginFromUnknownIP"], cyber_rules, "IntrusionDetected")
run_test("Cyber Scenario 3", ["MultipleFailedLogins", "LoginFromUnknownIP", "AdminPrivileges"], cyber_rules, "IntrusionDetected")
```
