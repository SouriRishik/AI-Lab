### Experiment 1: Breadth-First Search (BFS) for Maze Navigation

Problem Statement:
Consider a maze represented as a graph, where each node corresponds to a location in the maze and 
edges represent valid paths between locations.
An agent is placed at a start node and must reach a goal node by traversing the maze.

Task:
1. Formulate the maze navigation problem as a search problem by clearly defining:
- State space
- Initial state
- Actions
- Transition model
- Goal test
- Path cost
2. Implement the **Breadth-First Search (BFS)** algorithm to find the shortest path from the start node to the goal node.
3. Print the path returned by BFS.
4. Count and display the number of nodes expanded during the search.

Constraints:
- The maze is unweighted.
- The environment is fully observable, deterministic, static, discrete, and single-agent.
- BFS must be implemented without using any built-in graph search libraries.

```python
from collections import deque

def bfs(graph, start, goal):
    
    queue = deque()
    queue.append([start])
    
    visited = set()
    
    nodes_expanded = 0
    
    while queue:
        path = queue.popleft()
        current_node = path[-1]
        
        if current_node == goal:
            return path, nodes_expanded
        
        if current_node not in visited:
            visited.add(current_node)
            nodes_expanded += 1
            
            for neighbor in graph[current_node]:
                new_path = path + [neighbor]
                queue.append(new_path)
                
    return None, nodes_expanded

maze = {
    'A' : ['B', 'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E', 'G'],
    'G' : [],
}

start_node = 'A'
goal_node = 'G'

path, expanded = bfs(maze, start_node, goal_node)

print("Path found: ", path)
print("Nodes expanded: ", expanded)
```

### Experiment 2: Depth-First Search (DFS) for Graph Traversal

Problem Statement
Given a connected, finite graph represented using an adjacency list, design and implement the **Depth-First Search (DFS)** algorithm to find a path from a specified start node to a goal node.

Task Requirements
1. Formulate the graph traversal problem by identifying:
- State representation
- Initial state
- Goal state
- Actions

2. Implement a recursive DFS algorithm that:
- Explores one branch of the search tree as deeply as possible before backtracking
- Avoids revisiting nodes already present in the current path

3. Return and display:
- A valid path from the start node to the goal node (if it exists)

Constraints:
- The graph is unweighted.
- DFS must be implemented without using any built-in graph traversal libraries.
- The algorithm is not required to return the shortest path.

Expected Outcome
The algorithm should successfully find a path from the start node to the goal node by exploring depth-
wise, though the path may not be optimal.

```python
def dfs(graph, current, goal, path):
    expanded = 0
    expanded += 1
    path.append(current)
    if current == goal:
        return path
    for neighbor in graph[current]:
        if neighbor not in path:
            result = dfs(graph, neighbor, goal, path)
            if result is not None:
                return result
    path.pop()
    return None

graph = {
    'A' : ['B', 'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E', 'G'],
    'G' : [],
}


start_node = 'A'
goal_node = 'F'

path = dfs(graph, start_node, goal_node, [])

print("Path found: ", path)
print("Nodes expanded: ", expanded)
```

### Experiment 3: Solving the 8-Puzzle Problem Using BFS

Problem Statement:
The 8-puzzle problem consists of a 3×3 grid containing eight numbered tiles and one blank space. A tile adjacent to the blank space can be moved into the blank position.
Given an initial configuration and a goal configuration, implement **Breadth-First Search (BFS)** to find a sequence of moves that transforms the initial state into the goal state.

Task Requirements:
1. Represent the puzzle state using a suitable data structure.
2. Define:
- Initial state
- Goal state
- Valid actions (up, down, left, right)
- Transition model
3. Implement BFS to explore the state space.
4. Output:
- The sequence of states from the initial state to the goal state
- The total number of states expanded during the search
Constraints:
- Use BFS to guarantee the shortest solution.
- Avoid revisiting previously explored states.
- Assume the given initial state is solvable.

Expected Outcome:
The algorithm should produce the shortest sequence of moves required to reach the goal
configuration.

```python
import collections
import random

def get_moves(state):
    moves = []
    idx = state.index(0)
    row, col = divmod(idx, 3)
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for dr, dc in directions:
        nr, nc = row + dr, col + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_idx = nr * 3 + nc
            new_state = list(state)
            new_state[idx], new_state[new_idx] = new_state[new_idx], new_state[idx]
            moves.append(tuple(new_state))
    return moves

def is_solvable(state):
    flat = [n for n in state if n != 0]
    inversions = sum(flat[i] > flat[j] for i in range(len(flat)) for j in range(i + 1, len(flat)))
    return inversions % 2 == 0

def solve_bfs(initial_state, goal_state):
    queue = collections.deque([(initial_state, [initial_state])])
    visited = {initial_state}
    expanded_count = 0
    while queue:
        current_state, path = queue.popleft()
        expanded_count += 1
        if current_state == goal_state:
            return path, expanded_count
        for neighbor in get_moves(current_state):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    return None, expanded_count

def print_board(state):
    for i in range(0, 9, 3):
        row = state[i:i+3]
        print(f"{row[0]} {row[1]} {row[2]}")
    print()

if __name__ == "__main__":
    goal = (0, 1, 2, 3, 4, 5, 6, 7, 8)
    state_list = list(goal)
    while True:
        random.shuffle(state_list)
        initial = tuple(state_list)
        if is_solvable(initial): break

    solution_path, total_expanded = solve_bfs(initial, goal)
    print("Initial State:")
    print_board(initial)
    print("Sequence of states:")
    for step in solution_path:
        print_board(step)
    print(f"Total states expanded: {total_expanded}")
    print(f"Total moves: {len(solution_path) - 1}")
```

### Experiment 4: Performance Comparison of BFS and DFS

Problem Statement:
Breadth-First Search (BFS) and Depth-First Search (DFS) exhibit different performance characteristics in terms of time complexity, space complexity, and solution optimality.
Design an experiment to compare BFS and DFS when applied to the same search problem.

Task Requirements:
1. Apply both BFS and DFS to a common problem instance (graph traversal or maze navigation).
2. Instrument both algorithms to record:
- Number of nodes expanded
- Execution time
- Maximum memory used (queue or recursion depth)3. Tabulate the results obtained for BFS and DFS.
4. Analyze and explain the observed differences.

Constraints:
- Use the same start and goal states for both algorithms.
- Run each algorithm under identical conditions.
- Use Python’s time module for execution time measurement.

Expected Outcome:

Students should observe that:
- BFS guarantees the shortest path but uses more memory
- DFS uses less memory but may find non-optimal paths

```python
import collections, time, random

def solve(start, goal, mode, limit=15):
    start_time = time.time()
    container = collections.deque([(start, [start])])
    visited = {start: 0}
    expanded, max_mem = 0, 0

    while container:
        max_mem = max(max_mem, len(container))
        current, path = container.popleft() if mode == 'BFS' else container.pop()
        if len(path) > limit: continue
        expanded += 1
        if current == goal: return len(path)-1, expanded, time.time()-start_time, max_mem
        
        idx = current.index(0)
        r, c = divmod(idx, 3)
        for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
            nr, nc = r+dr, c+dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_idx = nr*3 + nc
                nxt = list(current); nxt[idx], nxt[new_idx] = nxt[new_idx], nxt[idx]
                nxt = tuple(nxt)
                if nxt not in visited or len(path) < visited[nxt]:
                    visited[nxt] = len(path)
                    container.append((nxt, path + [nxt]))
    return "Fail", expanded, time.time()-start_time, max_mem

goal = (0, 1, 2, 3, 4, 5, 6, 7, 8)
while True:
    initial = list(goal); random.shuffle(initial); initial = tuple(initial)
    flat = [n for n in initial if n != 0]
    inv = sum(flat[i] > flat[j] for i in range(8) for j in range(i+1, 8))
    res_b = solve(initial, goal, 'BFS')
    if inv % 2 == 0 and res_b[0] != "Fail" and res_b[0] < 12: break

print(f"Initial: {initial}\n{'Metric':<20} | {'BFS':<10} | {'DFS':<10}\n" + "-"*45)
for i, m in enumerate(["Path", "Nodes", "Time", "Memory"]):
    print(f"{m:<20} | {str(res_b[i])[:7]:<10} | {str(solve(initial, goal, 'DFS')[i])[:7]:<10}")
```

## Post-Exercise Analysis Questions

### 1. Why does BFS guarantee an optimal solution while DFS does not?

BFS explores nodes layer by layer, moving to a deeper level only after all nodes at the current depth are exhausted. This ensures the first time the goal is reached, it is via the shortest path. DFS follows one branch as deep as possible and may find a much longer path to the goal while missing a shallow solution in a different branch.

### 2. In which scenarios is DFS preferred over BFS?

DFS is preferred when **memory is highly constrained**, as it only needs to store the nodes on the current path (linear space). It is also useful in search spaces where solutions are dense and located deep in the tree, or when you need to explore a complete path to its conclusion (like in game trees or specific permutations).

### 3. How does the branching factor affect BFS performance?

The branching factor (\(b\)) causes the memory and time requirements of BFS to grow **exponentially**. For a solution at depth \(d\), BFS must store \(O(b^{d})\) nodes. Even a small increase in the branching factor significantly multiplies the number of states that must be kept in the queue. 

### 4. Why is BFS considered impractical for large state spaces?

BFS is impractical because its **space complexity** is its greatest bottleneck. Since it must keep every node it encounters in memory to maintain the frontier and track visited states, it will typically exhaust a computer's RAM long before it exhausts the available processing time in problems with high depth or high branching factors.
